欧几里得

```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

#测试三组
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))

#测试负数
print(gcd(-12, 8))
```

近似欧拉数

```python
import math

#1 euler_limit 极限形式 (1+1/n)^n
def euler_limit(n):
    return (1 + 1.0/n)**n

#2 euler_infinite_sum 前n项求和
def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k>0:
            fact *= k
        total += 1.0 / fact
    return total

#3 euler_approx，误差小于epsilon
def euler_approx(eps):
    true_e = math.exp(1)
    k = 0
    current_sum =0.0
    fact =1
    while abs(true_e - current_sum) >= eps:
        if k>0:
            fact *=k
        current_sum += 1.0/fact
        k +=1
    return current_sum

#4 print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1,n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

#测试
print(euler_infinite_sum(4))
print_euler_sum_table(5)
```

判断2的幂次

```python
def is_power_of_2(n):
    while n >1:
        if n %2 !=0:
            return False
        n = n//2
    return True

def is_power(b,n):
    while n >1:
        if n % b !=0:
            return False
        n = n //b
    return True

print(is_power_of_2(8))
print(is_power(3,81))
```
