字符串的乐趣

```python
# 1. 回文判断 is_palindrome
def is_palindrome(s: str) -> bool:
    clean = ''.join(s.split()).lower()
    return clean == clean[::-1]

# 2. 字母顺序 is_abecedarian：忽略空格，字母升序允许重复
def is_abecedarian(s: str) -> bool:
    clean = ''.join(s.split()).lower()
    for i in range(len(clean)-1):
        if clean[i] > clean[i+1]:
            return False
    return True

#3. remove_vowels 去除元音 a e i o u，大小写都去掉
def remove_vowels(s: str) -> str:
    vowels = {'a','e','i','o','u','A','E','I','O','U'}
    return ''.join(ch for ch in s if ch not in vowels)
# 测试
print(is_palindrome("T A C O cat"))  
print(is_abecedarian("abcd efgh xyz"))
print(is_palindrome("Was it a car or a cat I saw?")) 
print(remove_vowels('cat'))         
print(remove_vowels('audio'))        
```

列表的乐趣

```python
#1 list_reverse，输入不是列表抛错误
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是list列表类型")
    return lst[::-1]

#2 is_sorted，序列升序，只遍历一次
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

print(list_reverse([1,2,3]))  # [3,2,1]
print(is_sorted([1,2,4,5]))   # True
print(is_sorted([1,3,2]))     # False
```

作为向量的元组

```python
#1 vec_scalar_mult 元组向量乘标量，类型检查抛TypeError
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是int或float")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素必须是整数或浮点数")
    return tuple(x * s for x in t)

#2 vec_inner_product向量内积，元组检查，长度不等抛ValueError
def vec_inner_product(v1, v2):
    if not isinstance(v1, tuple) or not isinstance(v2, tuple):
        raise TypeError("输入必须是元组")
    if len(v1) != len(v2):
        raise ValueError("两个向量元组长度必须相等")
    for elem in v1+v2:
        if not isinstance(elem,(int,float)):
            raise TypeError("向量元素必须是数字")
    return float(sum(a*b for a,b in zip(v1, v2)))

#3 check_valid_mx 判断元组的元组是否合法矩阵
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    row_len = None
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if row_len is None:
            row_len = len(row)
        else:
            if len(row) != row_len:
                return False
        for val in row:
            if not isinstance(val, (int, float)):
                return False
    return True

#4 mx_vec_mult 矩阵（元组元组）乘向量（元组）
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("输入不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    ncol = len(mx[0])
    if len(vec) != ncol:
        raise ValueError("矩阵列数和向量维度不一致")
    res = []
    for row in mx:
        res.append(vec_inner_product(row, vec))
    return tuple(res)

#测试
print(vec_scalar_mult((1,2,3),2))          # (2,4,6)
print(vec_inner_product((1,2),(3,4)))      # 11.0
mat = ((1,2),(3,4))
print(check_valid_mx(mat))                 # True
print(mx_vec_mult(mat, (1,1)))            # (3.0,7.0)
```

更多向量的乐趣：Vector 类

```python
class Vector:
    def __init__(self, dim, elements=None):
        # 维度检查
        if not isinstance(dim, int):
            raise TypeError("维度dim必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        self.dim = dim
        # elements默认：不给元素就全部0
        if elements is None:
            self.elements = [0.0]*dim
        else:
            self.elements = list(elements)
            if len(self.elements) != dim:
                raise ValueError(f"元素个数{len(self.elements)}与维度{dim}不匹配")
            for num in self.elements:
                if not isinstance(num, (int,float)):
                    raise TypeError("向量元素必须是int/float")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个Vector维度不同")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个Vector维度不同")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个Vector维度不同")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个Vector维度不同")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("比较的两个Vector维度不同")
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不一致，无法点积")
        return float(sum(a*b for a,b in zip(self.elements, other.elements)))

    def __mul__(self, other):
        # v * scalar  或者 v * w（逐元素相乘）
        if isinstance(other, (int, float)):
            new_ele = [x*other for x in self.elements]
            return Vector(self.dim, new_ele)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("逐元素相乘，向量维度要相等")
            new_ele = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_ele)
        else:
            return NotImplemented

    def __rmul__(self, scalar):
        # scalar * v ，支持左乘
        return self * scalar

    def __repr__(self):
        return f"Vector({self.dim}, {tuple(self.elements)})"

# 测试Vector类
v1 = Vector(2,(2,4))
v2 = Vector(2,(2,5))
print(v1 < v2)          # True
print(2*v1)             # Vector(2, (4.0, 8.0))
print(v1*3)             # Vector(2, (6.0,12.0))
v3 = Vector(2,(2,3))
v4 = Vector(2,(4,5))
print(v3*v4)            # Vector(2, (8.0,15.0))
print(v3.dot(v4))       # 23.0
```
