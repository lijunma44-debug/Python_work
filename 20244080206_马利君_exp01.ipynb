实验二:函数,提交,循环,递归

```python
#1 score_level
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(82))
print(score_level(70))
print(score_level(50))
# score_level(105) #打开会抛异常

#2 for while求1~100和
sum_for = 0
for i in range(1,101):
    sum_for += i

sum_while = 0
i=1
while i <=100:
    sum_while += i
    i +=1
print("for求和1‑100：", sum_for)
print("while求和1‑100：", sum_while)

#3 迭代阶乘 & 递归阶乘
def fact_iter(n):
    res = 1
    for i in range(2, n+1):
        res *=i
    return res

def fact_recur(n):
    if n ==0:
        return 1
    return n * fact_recur(n-1)

for n in [0,1,5,10]:
    print(f"n={n},迭代={fact_iter(n)},递归={fact_recur(n)}")

#4 summary 返回最小最大平均，元组解包
def summary(values):
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)

vals = [10,20,30,40]
mi,ma,av = summary(vals)
print("解包结果：", mi, ma, av)

#5 sum_all 可变参数
def sum_all(*args):
    return sum(args)

print("无参", sum_all())
print("单参", sum_all(10))
print("多参", sum_all(1,2,3,4))
```

cell4‑1：目录创建，写 students.txt

```python
import os

#1 查看工作目录，创建文件夹
print("当前工作目录：", os.getcwd())
data_dir = "exp01_data"
os.makedirs(data_dir, exist_ok=True)
txt_path = os.path.join(data_dir, "students.txt")

#2 写入文本
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(txt_path, "w", encoding="utf‑8") as f:
    f.write(content)

#读取
with open(txt_path, "r", encoding="utf‑8") as f:
    lines = f.readlines()
print(lines)
```

cell4‑2：parse_student 函数 + 文件逐行处理

```python
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError("字段数量不对")
    name, score_str = parts
    name = name.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    score = int(score_str.strip())
    if not (0 <= score <= 100):
        raise ValueError(f"成绩{score}超出0‑100")
    return {"name":name, "score":score}

#逐行处理，捕获异常
valid_records = []
invalid_info = []

with open(txt_path, "r", encoding="utf‑8") as f:
    all_lines = f.readlines()

for lineno, line in enumerate(all_lines, start=1):
    if lineno == 1:
        continue #跳过表头
    try:
        rec = parse_student(line)
        valid_records.append(rec)
    except Exception as e:
        invalid_info.append({"line":lineno, "err":str(e)})

print("有效记录：", valid_records)
print("无效记录：", invalid_info)
valid_num = len(valid_records)
invalid_num = len(invalid_info)

#统计有效记录
if valid_num>0:
    avg_score = sum(r["score"] for r in valid_records)/valid_num
    pass_num = sum(1 for r in valid_records if r["score"]>=60)
    pass_r = pass_num / valid_num
else:
    avg_score=0
    pass_r=0

report_path = os.path.join(data_dir, "report.txt")

# w模式覆盖
with open(report_path, "w", encoding="utf‑8") as f:
    f.write(f"有效记录数:{valid_num}\n")
    f.write(f"无效记录数:{invalid_num}\n")
    f.write(f"平均分:{avg_score:.2f}\n")
    f.write(f"及格率:{pass_r:.1%}\n")

# a模式追加
with open(report_path, "a", encoding="utf‑8") as f:
    f.write("---追加测试(a模式)---\n")

print("w覆盖完成；a追加完成，打开report.txt查看")
```

实验五：类、继承、深浅拷贝

```python
import copy

#1 Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >=60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

#2 根据实验4有效记录创建对象
stu_objs = [Student(r["name"], r["score"]) for r in valid_records]
for s in stu_objs:
    print(s, "是否及格", s.is_passed())

#3 GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic})"

g_stu = GraduateStudent("LiHua",88,"AI")
print(g_stu, g_stu.is_passed())

#4 直接赋值、浅拷贝copy.copy、深拷贝copy.deepcopy
original = {"name":"Alice", "scores":[85,90]}

#①直接赋值
d1 = original
d1["scores"][0]= 999
print("直接赋值修改内层：orig", original, "d1", d1)
original = {"name":"Alice", "scores":[85,90]} #复原

#②浅拷贝
d2 = copy.copy(original)
d2["scores"][0]= 999
print("copy.copy浅拷贝修改内层：orig", original, "d2", d2)
original = {"name":"Alice", "scores":[85,90]} #复原

#③深拷贝
d3 = copy.deepcopy(original)
d3["scores"][0]=999
print("copy.deepcopy深拷贝修改内层：orig", original, "d3", d3)
```
