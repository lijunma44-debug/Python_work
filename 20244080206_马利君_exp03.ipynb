实验 1：正则表达式基础匹配
1. 匹配数字、空白字符和完整单词 cat

```python
import re
text = "cat on mat; later: cats and dogs"

patterns = {
    r"\d+": "数字",
    r"\s+": "空白字符",
    r"\bcat\b": "完整单词 cat"
}

for pattern, description in patterns.items():
    matches = re.findall(pattern, text)
    print(f"{description}，正则表达式 {pattern}：")
    print(matches)
    print()
```

2. 使用 cat|dog，比较不同分支顺序

```python
text = "cat dog bird CAT"

matches = re.findall(r"cat|dog", text)
print("cat|dog 的匹配结果：", matches)

text_a = "aaaa"

matches_1 = re.findall(r"a|aa|aaa", text_a)
matches_2 = re.findall(r"aaa|aa|a", text_a)

print("正则表达式 a|aa|aaa：", matches_1)
print("正则表达式 aaa|aa|a：", matches_2)
```

3. 比较不同数量词

```python
text = "ct cat caat caaat"

patterns = {
    r"ca*t": "ca*t：a 出现 0 次或多次",
    r"ca+t": "ca+t：a 出现 1 次或多次",
    r"ca?t": "ca?t：a 出现 0 次或 1 次",
    r"ca{2}t": "ca{2}t：a 恰好出现 2 次",
    r"ca{1,2}t": "ca{1,2}t：a 出现 1 到 2 次"
}

for pattern, description in patterns.items():
    matches = re.findall(pattern, text)
    print(description)
    print(f"{pattern} -> {matches}")
    print()
```

实验 2：分组、回溯引用与替换
1. 提取邮箱的用户名和域名

```python
text = "联系：alice@example.com；备选：bob@test.org"

email_pattern = r"([A-Za-z0-9._%+-]+)@([A-Za-z0-9.-]+\.[A-Za-z]{2,})"

matches = re.findall(email_pattern, text)

for username, domain in matches:
    print(f"用户名：{username}")
    print(f"域名：{domain}")
    print()
```

2. 使用后向引用找出连续重复的单词

```python
text = "cat cat, dog cat, bird bird"

# \b 表示单词边界，(\w+) 捕获一个单词，\1 表示重复前面捕获的内容
repeat_pattern = r"\b(\w+)\s+\1\b"

matches = re.findall(repeat_pattern, text)

print("连续重复的单词：", matches)
```

3. 将日期从“年-月-日”转换为“月-日-年”​

```python
text = """2024-05-01
2024/05/02
2024.05.03"""

print("转换前：")
print(text)

# 保留原来的分隔符
date_pattern = r"(\d{4})([-/.])(\d{2})(\2)(\d{2})"

# 月-日-年，并且使用原来的分隔符
replacement = r"\3\2\5\4\1"
converted_text = re.sub(date_pattern, replacement, text)

print("\n转换后：")
print(converted_text)
```

4. 统一电话号码格式

```python
text = "联系我：123-456-7890 或 987.654.3210"

phone_pattern = r"(\d{3})[-.](\d{3})[-.](\d{4})"
phone_replacement = r"(\1) \2-\3"

converted_phone_text = re.sub(
    phone_pattern,
    phone_replacement,
    text
)

print("转换前：")
print(text)

print("\n转换后：")
print(converted_phone_text)
```

实验 3：使用 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html>
<head>
    <title>实验课程</title>
</head>
<body>
    <h1>数据分析</h1>
    <p class="course">第六讲：正则表达式</p>
    <p class="course">第七讲：结构化数据</p>
    <a href="https://example.org/lecture06">讲义六</a>
    <a href="https://example.org/lecture07">讲义七</a>
</body>
</html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
```

1. 输出页面标题和 h1 信息

```python
title_text = soup.title.get_text()
h1_tag_name = soup.h1.name
h1_text = soup.h1.get_text()

print("页面标题：", title_text)
print("h1 标签名：", h1_tag_name)
print("h1 文本：", h1_text)
```

2. 查找课程段落和所有链接

```python
course_paragraphs = soup.find_all("p", class_="course")
links = soup.find_all("a")

print("所有课程段落：")
for index, paragraph in enumerate(course_paragraphs, start=1):
    print(f"{index}. {paragraph.get_text()}")

print("\n所有链接：")
for index, link in enumerate(links, start=1):
    print(f"{index}. 链接文字：{link.get_text()}")
    print(f"   href：{link.get('href')}")
```

3. 使用 get_text() 提取不同范围的文本

```python
whole_text = soup.get_text(separator="|", strip=True)
paragraph_text = course_paragraphs[0].get_text(strip=True)
link_text = links[0].get_text(strip=True)

print("整个文档的可见文本：")
print(whole_text)

print("\n第一个段落的文本：")
print(paragraph_text)

print("\n第一个链接的文本：")
print(link_text)
```
